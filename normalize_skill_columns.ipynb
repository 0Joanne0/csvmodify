## Normalize `Hard_Skills`, `Soft_Skills`, `Benefits`\n\nThis notebook runs the repo script `normalize_skill_columns.py` on your CSV and writes a new CSV with additional normalized columns:\n\n- `Hard_Skills_Normalized`\n- `Soft_Skills_Normalized`\n- `Benefits_Normalized`\n\nNotes:\n- The input file is semicolon-separated (`;`).\n- Items inside the skill columns are split on commas.\n- Normalization is case/accents/punctuation insensitive and deduplicates permutations.\n

```python
from pathlib import Path\nimport subprocess\n\ninput_csv = Path("data_jobs.csv")\noutput_csv = Path("data_jobs_normalized.csv")

hard_map = Path("hard_map.json")
soft_map = Path("soft_map.json")\n\nsubprocess.check_call([\n    "python3",\n    "normalize_skill_columns.py",\n    "--input", str(input_csv),\n    "--output", str(output_csv),
    "--hard-map", str(hard_map),
    "--soft-map", str(soft_map),\n])\n\noutput_csv\n
```

```python
import csv\n\nwith output_csv.open("r", encoding="utf-8", newline="") as f:\n    r = csv.DictReader(f, delimiter=";")\n    for i, row in enumerate(r):\n        if i >= 5:\n            break\n        print("Hard:", row.get("Hard_Skills"), "=>", row.get("Hard_Skills_Normalized"))\n        print("Soft:", row.get("Soft_Skills"), "=>", row.get("Soft_Skills_Normalized"))\n        print("Ben :", row.get("Benefits"), "=>", row.get("Benefits_Normalized"))\n        print("---")\n
```
